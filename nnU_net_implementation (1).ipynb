```python
from pathlib import Path

# adjust this if your path changes
labels_dir = Path("/content/ACDC/nnUNet_raw_data/Task501_ACDC/labelsTr")

if not labels_dir.is_dir():
    raise RuntimeError(f"{labels_dir} is not a directory")

for f in labels_dir.iterdir():
    if f.is_file() and "_gt" in f.name:
        new_name = f.name.replace("_gt", "")
        print(f"{f.name} → {new_name}")
        f.rename(labels_dir / new_name)
```

```python
from pathlib import Path
import shutil

src = Path("/content/ACDC/nnUNet_raw_data/Task501_ACDC/imagesTs")
dst = src.parent / "imagesTs_unique"
dst.mkdir(exist_ok=True)

patient_frames = {}
for f in src.glob("patient*_frame*.nii.gz"):
    base = f.name.removesuffix(".nii.gz")    # or f.name[:-7] if <3.9
    pid, frame = base.split("_frame")
    num = int(frame)
    if pid not in patient_frames or num < patient_frames[pid][0]:
        patient_frames[pid] = (num, f)

for pid, (_, f) in patient_frames.items():
    newname = f"{pid}_fr_0000.nii.gz"
    print(f"{f.name} → {newname}")
    shutil.copy(f, dst / newname)
```

```python
#!/usr/bin/env python3
import torch
from pathlib import Path

# adjust this base path if needed
base = Path("/content/nnUNet_results/nnUNet/3d_fullres/Task501_ACDC/nnUNetTrainerV2__nnUNetPlansv2.1")

for fold in range(5):
    model_file = base / f"fold_{fold}" / "model_final_checkpoint.model"
    if not model_file.is_file():
        print(f"Fold {fold}: {model_file} not found, skipping")
        continue

    print(f"Fold {fold}: loading → {model_file.name}")
    ckpt = torch.load(model_file, map_location="cpu", weights_only=False)

    print(f"Fold {fold}: re‑saving → {model_file.name}")
    torch.save(ckpt, model_file)

print("All done!")
```

```python
import nibabel as nib
import numpy as np
from skimage import measure
import plotly.graph_objects as go

# 1) Load the segmentation
pred_nii = nib.load("/content/ACDC_preds/patient101_fr.nii.gz")
seg   = pred_nii.get_fdata().astype(int)
spacing = pred_nii.header.get_zooms()  # voxel size (x,y,z)

# 2) Define which structures to mesh
labels = {
    "LV" : 1,   # left ventricle
    "RV" : 2,   # right ventricle
    "MYO": 3    # myocardium
}
colors = {"LV":"red", "RV":"blue", "MYO":"green"}

# 3) Create meshes
fig = go.Figure()
for name,lab in labels.items():
    mask = seg == lab
    if mask.sum()==0: continue
    verts, faces, _, _ = measure.marching_cubes(
        mask, level=0.5, spacing=spacing)
    fig.add_trace(
      go.Mesh3d(
        x=verts[:,0], y=verts[:,1], z=verts[:,2],
        i=faces[:,0], j=faces[:,1], k=faces[:,2],
        color=colors[name], opacity=0.5,
        name=name
      )
    )

# 4) Show
fig.update_layout(
  title="3D Reconstruction of Patient101",
  scene=dict(aspectmode="data",
             xaxis_visible=False,
             yaxis_visible=False,
             zaxis_visible=False)
)
fig.show()
```
